# Determination of the Specific Heat Capacity of Water


# Introduction

Water has an enormous heat buffering capacity: a significant amount of energy is required to raise its temperature by just one degree. In this lab, we will determine the specific heat capacity of water by heating a known mass of water with a measured amount of electrical energy and recording the resulting temperature rise.

# Theory

The specific heat capacity ($c$) is the energy required to raise the temperature of 1 gram of a substance by 1°C. The heat absorbed by the water ($Q$) is given by:

$$Q = m \cdot c \cdot \Delta T$$

Where:
- $m$ is the mass (g)
- $c$ is the specific heat (J/g°C)
- $\Delta T$ is the change in temperature ($T_{final} - T_{initial}$)

In this experiment, the energy is provided by an electrical heater. The electrical energy ($E$) supplied is:

$$E = V \cdot I \cdot t$$

Assuming an ideal system with no heat loss, $E = Q$. Therefore:

$$c = \frac{V \cdot I \cdot t}{m \cdot \Delta T}$$


# Methods and Materials

## Design
A water bath of known mass is heated using an electrical immersion heater. By measuring voltage, current, and the temperature rise over a fixed time interval, the specific heat capacity can be calculated.

## Materials
- Beaker
- Digital Scale
- Water
- Immersion Heater ($10 \Omega$, $10 W$)
- DC Power Supply
- Digital Thermometer
- Stopwatch

## Procedure
1. Measure the mass of the empty beaker and the mass after adding ~250g of water to find the water mass ($m_w$).
2. Submerge the heater and thermometer in the water.
3. Record the initial temperature ($T_1$).
4. Turn on the power supply and start the timer.
5. Maintain a constant Voltage ($V$) and Current ($I$) for 600 seconds.
6. Stir continuously to ensure even heat distribution.
7. Record the final temperature ($T_2$) and calculate $\Delta T$.

## Data Analysis
We will calculate the total electrical work done ($E$) and divide it by the product of the water's mass and temperature change to find $c$.

# Results


```python
import matplotlib.pyplot as plt

# Input Data
m_w = 250.0  # grams
V = 12.0     # Volts
I = 3.0      # Amperes
t = 600      # Seconds
T1 = 22.0    # Celsius
T2 = 40.5    # Celsius

# Calculations
E = V * I * t
delta_T = T2 - T1
c_exp = E / (m_w * delta_T)

print(f"Energy Input: {E} Joules")
print(f"Temp Change: {delta_T} °C")
print(f"Experimental Specific Heat: {c_exp:.2f} J/g°C")

# Plotting result
plt.plot([0, t], [T1, T2], 'bo-')
plt.title('Temperature vs Time')
plt.xlabel('Time (s)')
plt.ylabel('Temperature (°C)')
plt.grid(True)
plt.show()
```

# Discussion and Conclusion

**Conclusion:**
The experimental value for the specific heat of water was found to be $4.67 \text{ J/g°C}$. This is higher than the standard value of $4.18 \text{ J/g°C}$.

**Discussion:**
The primary source of error is heat loss to the surroundings and the heat absorbed by the calorimeter (beaker). Since $E$ is constant, a lower-than-expected $T_2$ (due to heat loss) results in an artificially high value for $c$. Using a lid and insulation would improve accuracy.
