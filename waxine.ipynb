# Waxinekaarsje

Als je een waxinekaarsje aan steekt en er een glas overheen zet, zal het kaarsje uit gaan. 
Als je dat waxinekaarsje laat drijven op water terwijl je er het glas er overheen zet, zal het water in het glas stijgen.
Op het internet zwerven er verschillende verklaringen voor dit fenomeen rond.
1. Het kaarsje verbruikt zuurstof, waardoor er een vacuüm ontstaat en het water omhoog wordt gezogen.
2. Het kaarsje verwarmt de lucht in het glas, waardoor de lucht uitzet. Wanneer de kaars dooft koelt de lucht af waardoor de druk afneemt en er water omhoog wordt gezogen.
3. Bij het verbranden van de kaars ontstaan er waterdamp en koolstofdioxide. De waterdamp condenseert aan de binnenkant van het glas, waardoor er minder gas in het glas is en de druk afneemt. Hierdoor wordt het water omhoog gezogen.

```{exercise}
Onderzoek welke verklaring het meest waarschijnlijk is en leg uit waarom de andere verklaringen minder waarschijnlijk zijn.
```

# Analyse Oefening 1: De fysica achter het stijgende waterniveau

In dit experiment onderzoeken we drie verklaringen voor het stijgen van het water in een bekerglas nadat dit over een brandende kaars is geplaatst. Op basis van de verzamelde meetgegevens en de wetten van de thermodynamica volgt hier de analyse.

---

## 1. Beoordeling van de verklaringen

### Hypothese 1: Zuurstofverbruik — **Onwaarschijnlijk als hoofdoorzaak**
De kaars gebruikt wel zuurstof, maar dat is niet de hoofdreden dat het water stijgt.
* **Chemische substitutie:** Bij de verbranding van paraffine wordt zuurstof ($O_{2}$) omgezet in koolstofdioxide ($CO_{2}$) en waterdamp ($H_{2}O$). Omdat gasmoleculen elkaar grotendeels vervangen, ontstaat er door de chemische reactie alleen geen significant vacuüm.
* **Volumebeperking:** Zuurstof beslaat theoretisch ongeveer 21% van het luchtvolume. Uit de data blijkt dat het volume van het gestegen water ($V_{water}$) vaak aanzienlijk afwijkt van deze 21%-drempel.

### Hypothese 2: Thermische uitzetting en contractie — **Meest waarschijnlijke oorzaak**
De thermodynamica levert de sterkste verklaring voor het waargenomen fenomeen.
* **Expansie:** De brandende kaars verhit de lucht in de beker, waardoor deze uitzet (Wet van Charles). Tijdens de verwarmingsfase zijn er bellen waargenomen die onder de rand van de beker ontsnapten, wat bewijst dat er luchtmassa uit het systeem verdween.
* **Contractie:** Zodra de vlam dooft door zuurstofgebrek, koelt het resterende gas snel af. Volgens de wet van Gay-Lussac ($P \propto T$) daalt de interne druk onder de atmosferische druk. Dit drukverschil dwingt het water omhoog tot de druk in de beker weer in evenwicht is met de buitenlucht.
* **Timing:** Het feit dat de waterstijging pas echt versnelt *nadat* de vlam gedoofd is, bevestigt dat afkoeling de drijvende kracht is.

### Hypothese 3: Condensatie van waterdamp — **Ondersteunende factor**
Deze verklaring draagt bij aan het drukverschil, maar is afhankelijk van de temperatuurverandering.
* **Faseovergang:** Bij verbranding ontstaat waterdamp. Wanneer deze damp afkoelt en condenseert tot vloeibaar water (zichtbaar als mist op het glas), neemt het ingenomen volume van deze moleculen drastisch af. Dit verlaagt de druk in de beker nog verder.

---

## 2. Experimentele Data en Berekeningen

De onderstaande tabel bevat de meetwaarden die zijn gebruikt om de volumes te bepalen, waarbij het volume van de kaars ($V_{candle} \approx 17.322,62 \text{ mm}^3$) is verrekend om het netto watervolume te vinden.

| Meeks | Straal ($r_{beaker}$) | Hoogte stijging ($h_{rise}$) | Netto watervolume ($V_{water}$) |
| :--- | :--- | :--- | :--- |
| **Beaker 1 (Small)** | 23.0 mm  | 26.0 mm  | 25.892,69 $mm^{3}$  |
| **Beaker 2 (Medium)** | 43.0 mm  | 22.0 mm  | 110.471,88 $mm^{3}$  |
| **Beaker 3 (Large)** | 24.5 mm  | 44.5 mm  | 66.589,81 $mm^{3}$  |



## 3. Eindconclusie
Op basis van de waarnemingen (ontsnappende bellen en de timing van de stijging) en de berekeningen (afwijking van de 21% zuurstof-limiet), concluderen we dat de **thermische contractie** (verklaring 2) in combinatie met **condensatie** (verklaring 3) de hoofdoorzaken zijn van de waterstijging. Hypothese 1 is fysiek onvoldoende om het volledige effect te verklaren.
